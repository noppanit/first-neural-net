## 1. Meet Python's data table tool: pandas

A **DataFrame** is a table like a spreadsheet. `pd` is just a short nickname for pandas.

Run the next cell. If it says `ModuleNotFoundError`, install the packages in `requirements.txt` first.

```python
import pandas as pd
from pathlib import Path

# Works when Jupyter starts in the repository root OR the notebooks folder.
path = Path('data/Advertising.csv')
if not path.exists():
    path = Path('../data/Advertising.csv')

df = pd.read_csv(path)
df = df.drop(columns=['Unnamed: 0'], errors='ignore')
print('Loaded the dataset!')
df.head()
```

## 2. What am I looking at?

Each **row** is one advertising-spending example. Each **column** is one measurement.

- **TV**, **Radio**, **Newspaper** = advertising budgets (our *features*, also called inputs)
- **Sales** = sales outcome (our *target*, the number we hope to predict later)

These figures are from an educational dataset, so don't treat them as real campaign results or causal proof.

**Your turn:** What would we be predicting if we built a model from this table?

```python
print('Rows and columns:', df.shape)
print('Column names:', df.columns.tolist())
print('\nData types:')
print(df.dtypes)
```

## 3. Is any data missing?

Real-world datasets often have blank cells. Let's count them. A `0` means none are missing from that column.

```python
df.isna().sum()
```

## 4. Basic statistics, without scary math

- **mean** = average
- **min** = smallest value
- **max** = largest value
- **std** = standard deviation: roughly how spread out values are

The next command calculates all of these for each column.

```python
df.describe().round(2)
```

## 5. Look at one column at a time

The square brackets select a column. Try changing `'TV'` to `'Radio'`.

```python
print('Average TV advertising:', round(df['TV'].mean(), 2))
print('Largest TV advertising:', df['TV'].max())
print('Smallest TV advertising:', df['TV'].min())
```
